# PyMongo Tutorial - Book Management System
## Basic MongoDB Operations with Python

## 1. Connect to MongoDB

```python
from pymongo import MongoClient
from dotenv import load_dotenv
import os

load_dotenv()
MONGODB_URI = os.getenv("MONGODB_URI")

# Connect to MongoDB Atlas
client = MongoClient(MONGODB_URI)
db = client.library
books = db.books2

print("✓ Connected to MongoDB!")
print(f"Database: {db.name}")
print(f"Collection: {books.name}")
```

## 2. CREATE - Insert Documents

```python
# Insert single book
book1 = {
    "title": "The Great Gatsby",
    "author": "F. Scott Fitzgerald",
    "genre": "Fiction",
    "year": 1925
}

result = books.insert_one(book1)
print(f"Book inserted with ID: {result.inserted_id}")
```

```python
# Insert multiple books
new_books = [
    {"title": "1984", "author": "George Orwell", "genre": "Dystopian", "year": 1949},
    {"title": "To Kill a Mockingbird", "author": "Harper Lee", "genre": "Fiction", "year": 1960},
    {"title": "Pride and Prejudice", "author": "Jane Austen", "genre": "Romance", "year": 1813}
]

results = books.insert_many(new_books)
print(f"✓ Inserted {len(results.inserted_ids)} books")
```

## 3. READ - Fetch Documents

```python
# Get all books
all_books = books.find()
print("All Books in Library:")
for book in all_books:
    print(f"  - {book['title']} by {book['author']} ({book['year']})")
```

```python
# Get single book
book = books.find_one({"title": "1984"})
print("\nFound Book:")
print(f"Title: {book['title']}")
print(f"Author: {book['author']}")
print(f"Genre: {book['genre']}")
```

```python
# Filter - Find all Fiction books
fiction_books = books.find({"genre": "Fiction"})
print("\nFiction Books:")
for book in fiction_books:
    print(f"  - {book['title']}")
```

## 4. UPDATE - Modify Documents

```python
# Update book genre
result = books.update_one(
    {"title": "1984"},
    {"$set": {"genre": "Political Fiction"}}
)
print(f"✓ Updated {result.modified_count} book")

# Verify update
updated = books.find_one({"title": "1984"})
print(f"New genre: {updated['genre']}")
```

## 5. DELETE - Remove Documents

```python
# Delete single book
result = books.delete_one({"title": "Pride and Prejudice"})
print(f"✓ Deleted {result.deleted_count} book")

# Check total books
total = books.count_documents({})
print(f"\nTotal books remaining: {total}")
```

## 6. Analytics

```python
# Get statistics
total_books = books.count_documents({})
unique_authors = books.distinct("author")
unique_genres = books.distinct("genre")

print(f"Total Books: {total_books}")
print(f"Authors: {unique_authors}")
print(f"Genres: {unique_genres}")
```

```python
# Convert to DataFrame for analysis
import pandas as pd

data = list(books.find({}, {"_id": 0}))
df = pd.DataFrame(data)
print("\nBooks DataFrame:")
print(df)
```

## Key PyMongo Methods Summary
- **insert_one()** - Add 1 document
- **insert_many()** - Add multiple documents
- **find()** - Get all documents (returns cursor)
- **find_one()** - Get first matching document
- **update_one()** - Modify 1 document with $set operator
- **delete_one()** - Remove 1 document
- **count_documents()** - Count documents
- **distinct()** - Get unique values for a field
