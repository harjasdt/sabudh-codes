```python
from ollama import chat
import json

MODEL = 'llama3.1:8B'

def call_llm(messages, temperature=0.7, model=MODEL):
    """Helper function to call the LLM"""
    response = chat(
        model=model,
        messages=messages,
        options={'temperature': temperature}
    )
    return response.message.content

# Test connection
test = call_llm([{'role': 'user', 'content': 'Say hello!'}])
print(f"Model loaded successfully!\nTest response: {test[:100]}...")
```

## 1. Zero-Shot Prompting

```python
prompt = "Classify the sentiment of this review: 'The product is amazing and works perfectly!'"
response = call_llm([{'role': 'user', 'content': prompt}])
print("ZERO-SHOT PROMPTING")
print("="*50)
print(f"Prompt: {prompt}")
print(f"\nResponse:\n{response}")
```

## 4. Chain-of-Thought (CoT) Prompting

```python
prompt = r"""Solve this step by step:
If a book costs $15 and you get a 20% discount, how much do you pay?

Think through each step:
1. Calculate the discount amount
2. Subtract from original price
3. Give final answer"""

response = call_llm([{'role': 'user', 'content': prompt}])
print("CHAIN-OF-THOUGHT PROMPTING")
print("="*50)
print(f"Response:\n{response}")
```

## 5. Few-Shot + Chain-of-Thought

```python
prompt = r"""Solve math problems step by step.

Example 1:
Problem: A shirt costs $30 with 10% discount. Final price?
Step 1: 10% of $30 = $3
Step 2: $30 - $3 = $27
Answer: $27

Example 2:
Problem: If you have 5 boxes with 12 items each, how many total?
Step 1: 5 boxes × 12 items per box
Step 2: 5 × 12 = 60
Answer: 60 items

Now solve:
Problem: A notebook costs $8, pens cost $2 each. You buy 1 notebook and 4 pens. Total cost?
Step 1:
Step 2:
Step 3:
Answer:"""

response = call_llm([{'role': 'user', 'content': prompt}])
print("FEW-SHOT + CHAIN-OF-THOUGHT")
print("="*50)
print(f"Response:\n{response}")
```

## 9. Structured Output Prompting (Pydantic)

Use Pydantic schemas to enforce JSON structure directly.

```python
from pydantic import BaseModel

# 1. Define your Pydantic model
class SentenceAnalysis(BaseModel):
    word_count: int
    unique_words: list[str]  # Use specific list typing for better validation
    longest_word: list[str]
    has_adjectives: bool

# 2. Simple prompt without raw formatting instructions
prompt = "Analyze this sentence: 'The quick brown fox jumps over the lazy dog.' word_count means individual words in the sentence and not characters. unique_words means the set of distinct words in the sentence. longest_word means the list of words with the most letters in the sentence. has_adjectives means whether there are any adjectives in the sentence."

# 3. Pass the Pydantic JSON schema directly to format parameter
response = chat(
    model=MODEL,
    messages=[{"role": "user", "content": prompt}],
    format=SentenceAnalysis.model_json_schema()  # Direct schema enforcement
)

print("STRUCTURED OUTPUT PROMPTING (Pydantic Schema)")
print("="*50)
print(f"Raw Response:\n{response.message.content}\n")

# 4. Parse using Pydantic's native helper
try:
    validated = SentenceAnalysis.model_validate_json(response.message.content)
    print("✓ Successfully parsed and validated!\n")
    print(f"Word Count: {validated.word_count}")
    print(f"Unique Words: {validated.unique_words}")
    print(f"Longest Word: {validated.longest_word}")
    print(f"Has Adjectives: {validated.has_adjectives}")
except Exception as e:
    print(f"Validation error: {e}")
```

## 12. Self-Consistency Prompting

```python
prompt = """A store has 20 apples. They sell 5 apples on Monday, buy 8 more on Tuesday, and sell 3 on Wednesday. How many apples are left? Give just the number."""

print("SELF-CONSISTENCY PROMPTING")
print("="*50)

for i in range(3):
    resp = call_llm([{'role': 'user', 'content': prompt}], temperature=0.7)
    print(f"Run {i+1}: {resp}")

print(f"\nCorrect: 20 - 5 + 8 - 3 = 20 apples")
```

| Limitation | Risk | Mitigation |
|-----------|------|------------|
| **Hallucination** | Makes up facts | Verify, ask for sources, low temperature |
| **Bias** | Stereotypes | Test diverse inputs, human review |

# SECTION 5: EMBEDDINGS & SEMANTIC SEARCH

## What are Embeddings?

Convert text to vectors that capture semantic meaning:
- Similar meaning → Similar vectors → Close in space
- We can do math on meaning

## King-Queen Analogy Demo

Classic embedding example: **king - man + woman = queen**

```python
!pip install -q sentence-transformers scikit-learn torch 
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Load model
model = SentenceTransformer('all-MiniLM-L6-v2')
words = ['king', 'queen', 'man', 'woman', 'prince', 'princess']
embeddings = model.encode(words)

print("EMBEDDINGS: KING-QUEEN ANALOGY")
print("="*50)
print(f"Dimensions: {embeddings.shape[1]}\n")

# king - man + woman = ?
king_emb = embeddings[0]
man_emb = embeddings[2]
woman_emb = embeddings[3]
result = king_emb - man_emb + woman_emb

print("king - man + woman = ?\n")
for i, word in enumerate(words):
    sim = cosine_similarity([result], [embeddings[i]])[0][0]
    print(f"  {word:10s}: {sim:.4f}")

print(f"\n✓ 'queen' has highest similarity!")
print(f"Embeddings understand gender relationships!\n")

# prince - man + woman = ?
prince_emb = embeddings[4]
result2 = prince_emb - man_emb + woman_emb

print("prince - man + woman = ?\n")
for i, word in enumerate(words):
    sim = cosine_similarity([result2], [embeddings[i]])[0][0]
    print(f"  {word:10s}: {sim:.4f}")

print(f"\n✓ 'princess' has highest similarity!")
```

# SECTION 6: BEST PRACTICES

## 8 Core Principles

1. **Start Simple** - Zero-shot before few-shot
2. **Be Specific** - Clear > vague
3. **Show Examples** - Few-shot power
4. **Use Structure** - JSON > plain text
5. **Chain Tasks** - Break into steps
6. **Control Temperature** - Low for facts, high for ideas
7. **Always Verify** - Especially critical tasks
8. **Keep Iterating** - No perfect prompt
